# Day 8: Vehicle Detection using COCO Dataset

We will fine-tune YOLO11 on the COCO dataset focusing on vehicle classes.
This is a strong real-world project (Traffic analysis, Smart City, etc.).

```python
# Load model
model = YOLO("yolo11s.pt")

# Train on COCO (it will auto-download)
results = model.train(
    data="coco8.yaml",
    epochs=30,
    imgsz=640,
    batch=16,
    name="vehicle_coco",
    patience=15,
    augment=True
)
```

```python
best_model = YOLO("runs/detect/vehicle_coco/weights/best.pt")

test_images = [
    "https://ultralytics.com/images/bus.jpg",
    "https://ultralytics.com/images/zidane.jpg"
]

for url in test_images:
    results = best_model(url, conf=0.3)
    results[0].show()
```

The results of the video detection are saved to a directory within `runs/detect/`. Let's list the contents of that directory to find the processed video.

```python
import os

# List the contents of the directory where results are saved
results_dir = '/content/runs/detect/predict'

# Check if the directory exists, as it might be 'predict' or 'predictN' for subsequent runs
if not os.path.exists(results_dir):
    # Find the most recent 'predict' directory if 'predict' itself doesn't exist
    predict_dirs = [d for d in os.listdir('/content/runs/detect') if d.startswith('predict')]
    predict_dirs.sort(key=lambda x: int(x.replace('predict', '') or 0))
    if predict_dirs:
        results_dir = os.path.join('/content/runs/detect', predict_dirs[-1])


print(f"Contents of {results_dir}:")
for item in os.listdir(results_dir):
    print(item)

# Optionally, you can display the video in Colab if it's a small file
# from IPython.display import Video
# Video(os.path.join(results_dir, 'downloaded_video.mp4'))
```
