# Day 17: Zero-Shot Object Detection using CLIP + YOLO

## Key New Concepts Learned Today

**Note 1: What is Zero-Shot Detection?**
- Traditional models can only detect classes they were trained on.
- Zero-Shot = You can detect **any object** by just typing its name (no training needed).

**Note 2: How it works**
- YOLO detects possible objects (boxes)
- CLIP (from OpenAI) compares the image region with your text description
- Very powerful and flexible

**Note 3: Why this is advanced?**
- One of the most modern techniques in 2026
- Extremely useful when you don't have labeled data
- Highly valued in research and industry

```python
# Load models
yolo_model = YOLO("yolo11s.pt")
clip_classifier = pipeline("zero-shot-image-classification",
                          model="openai/clip-vit-large-patch14")

def zero_shot_detection(image, candidate_labels):
    # Step 1: YOLO detects regions
    results = yolo_model(image, conf=0.3, verbose=False)

    annotated = results[0].plot()

    # Step 2: For each detected box, use CLIP for classification
    for i, box in enumerate(results[0].boxes):
        # Crop the detected region
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        cropped = image[y1:y2, x1:x2]

        # CLIP Zero-Shot Classification
        result = clip_classifier(Image.fromarray(cropped), candidate_labels)

        # Update label with highest confidence
        best_label = result[0]['label']
        best_score = result[0]['score']

        print(f"Object {i+1}: {best_label} ({best_score:.2f})")

    return annotated
```
