# Day 19: Model Quantization & Edge Deployment Optimization

## Key New Concepts Learned Today

**Note 1: What is Model Quantization?**
- Reducing the precision of model weights (from 32-bit float → 8-bit integer)
- Makes model much smaller and faster
- Slight accuracy trade-off

**Note 2: Why This Is Important?**
- Original YOLO model = Large & Slow
- Quantized model = Small, Fast, Can run on Mobile Phones, Raspberry Pi, Jetson Nano
- This is a **highly demanded skill** in industry

**Note 3: Types of Quantization**
- FP16 (Half Precision)
- INT8 (Most Popular for Edge)
- Dynamic Quantization

```python
# Load your best model
model = YOLO("yolov8n.pt")   # Changed from 'models/best.pt' to 'yolov8n.pt'

print("Original Model Size:", os.path.getsize("yolov8n.pt") / (1024*1024), "MB")

# 1. Export to ONNX
model.export(format="onnx")

# 2. INT8 Quantization (Most Important)
model.export(format="onnx", int8=True)

print("✅ Quantized model created!")
```

```python
print("=== Speed Benchmark ===")

# Original Model
%timeit model.predict("https://ultralytics.com/images/bus.jpg", conf=0.3, verbose=False)

# Quantized Model (after export)
quantized_model = YOLO("yolov8n_int8.onnx")
%timeit quantized_model.predict("https://ultralytics.com/images/bus.jpg", conf=0.3, verbose=False)
```
