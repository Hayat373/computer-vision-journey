```python
import cv2
from collections import defaultdict

model = YOLO("yolo11s.pt")

cap = cv2.VideoCapture(SOURCE_VIDEO_PATH)
count_dict = defaultdict(int)
frame_count = 0

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break

    frame_count += 1
    results = model.track(frame, persist=True, conf=0.3, tracker="bytetrack.yaml")

    for result in results:
        boxes = result.boxes
        for box in boxes:
            cls = int(box.cls[0])
            class_name = model.names[cls]
            if class_name in ['car', 'bus', 'truck', 'motorcycle']:
                count_dict[class_name] += 1

print("📊 Final Vehicle Count:")
for vehicle, count in count_dict.items():
    print(f"{vehicle.capitalize()}: {count}")

cap.release()
```
