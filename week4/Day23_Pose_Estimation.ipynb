```python
import cv2
import mediapipe as mp
from ultralytics import YOLO
import numpy as np
from IPython.display import Image, display
import matplotlib.pyplot as plt

print("✅ Libraries imported!")
```

```python
mp_pose = mp.solutions.pose
mp_drawing = mp.solutions.drawing_utils

# Load image or use webcam later
image_path = "https://ultralytics.com/images/bus.jpg"  # change later

# For image
cap = cv2.VideoCapture(image_path) if image_path.startswith("http") else cv2.imread(image_path)

with mp_pose.Pose(static_image_mode=True, min_detection_confidence=0.5) as pose:
    results = pose.process(cv2.cvtColor(cap, cv2.COLOR_BGR2RGB))
    
    if results.pose_landmarks:
        annotated_image = cap.copy()
        mp_drawing.draw_landmarks(
            annotated_image,
            results.pose_landmarks,
            mp_pose.POSE_CONNECTIONS
        )
        plt.figure(figsize=(10,8))
        plt.imshow(cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB))
        plt.title("MediaPipe Pose Estimation")
        plt.show()
```

```python
model = YOLO("yolo11s-pose.pt")   # Download automatically

results = model("https://ultralytics.com/images/bus.jpg", conf=0.5)

# Show result
results[0].show()

# Save output
results[0].save("pose_output.jpg")
```
