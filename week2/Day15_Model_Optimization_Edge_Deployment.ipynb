# Day 15: Model Optimization & Edge Deployment

## Key New Concepts Learned Today

**Note 1: Why Optimization?**
- Original YOLO model is large and slow
- Optimized model = faster inference, smaller size, runs on mobile/edge devices

**Note 2: Common Optimization Techniques**
- Export to ONNX format
- Quantization (INT8 / FP16)
- TensorRT (for NVIDIA GPUs)
- OpenVINO (for Intel CPUs)

**Note 3: Real-world Importance**
- Needed for mobile apps, Raspberry Pi, security cameras, drones, etc.
- Very valuable skill for jobs and hackathons

```python
# Load your best model
model_filename = "yolo11s.pt"
model = YOLO(model_filename)

print("Original Model Size:", os.path.getsize(model_filename) / (1024*1024), "MB")

# Export to different formats
print("\nExporting Model...")

# 1. ONNX (Most Popular)
model.export(format="onnx")

# 2. TensorRT (Best for NVIDIA)
# model.export(format="engine")

# 3. OpenVINO
model.export(format="openvino")


print("✅ Model exported successfully!")
```

```python
# Benchmark different formats
print("Speed Benchmark:")

# Original PyTorch
results = model.predict("https://ultralytics.com/images/bus.jpg", conf=0.3, verbose=False)
print("PyTorch Speed: Fast")

# ONNX version (usually faster)
onnx_model = YOLO("yolo11s.onnx")
results_onnx = onnx_model.predict("https://ultralytics.com/images/bus.jpg", conf=0.3, verbose=False)
print("ONNX Speed: Usually Faster")
```
